```python
import numpy as np
import pandas as pd
```

### Second Level

```python
df = pd.read_csv('/content/drive/MyDrive/ml/ECPICK/2_L/test_sequences.csv')
sequences = df[['sequence']]
```

```python
fasta_path = 'sample_sl.fasta'
with open(fasta_path, 'w') as f:
    for i, seq_record in enumerate(sequences['sequence']):
        # Assuming a simple header format for each sequence
        f.write(f">seq_{i}\n{seq_record}\n")

print(f"Archivo FASTA '{fasta_path}' creado exitosamente.")

# Ahora, usa ECPICK para predecir las secuencias
# Asegúrate de ejecutar este comando dentro del entorno 'py39'
# Los comandos de shell deben ser prefijados con !conda run -n py39
# Sin embargo, si quieres usar un objeto ECPICK directamente en Python, necesitarás importar y ejecutarlo dentro de un script de Python
# que luego se ejecute con !conda run -n py39 python -c "..."
# Para simplificar, generaré un script temporal y lo ejecutaré.

prediction_script = """
from ecpick import ECPICK

ecpick = ECPICK()
ecpick.predict_fasta(fasta_path='sample_sl.fasta', output_path='output_ecpick_sl')
print("ECPICK predictions initiated. Check 'output_ecpick_sl' directory for results.")
"""

# Save the prediction script to a file
with open('run_ecpick.py', 'w') as f:
    f.write(prediction_script)

# Execute the script using the py39 environment
!conda run -n py39 python run_ecpick.py
```
