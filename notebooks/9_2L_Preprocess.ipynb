```python
from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
current_dir = Path.cwd()
ptf_path = str(current_dir.parent / 'datasets/First Level/pre_processed/train_feat.npy')
pte_path = str(current_dir.parent / 'datasets/First Level/pre_processed/train_emb.npy')
pvf_path = str(current_dir.parent / 'datasets/First Level/pre_processed/valid_feat.npy')
pve_path = str(current_dir.parent / 'datasets/First Level/pre_processed/valid_emb.npy')
ptef_path = str(current_dir.parent / 'datasets/First Level/pre_processed/test_feat.npy')
ptee_path = str(current_dir.parent / 'datasets/First Level/pre_processed/test_emb.npy')

df_train_path = str(current_dir.parent / 'datasets/First Level/train_descriptores.csv')
df_valid_path = str(current_dir.parent / 'datasets/First Level/valid_descriptores.csv')
df_test_path = str(current_dir.parent / 'datasets/First Level/test_descriptores.csv')
```

```python
df_train = pd.read_csv(df_train_path)
df_valid = pd.read_csv(df_valid_path)
df_test = pd.read_csv(df_test_path)
```

```python
df = pd.concat([df_train, df_valid, df_test], ignore_index=True)
df.head(5)
```

```python
indexs = []
y = []
for i in range(len(df)):
    if (not df.iloc[i,2].startswith('0')):
        indexs.append(i)
        y.append(int(str(df.iloc[i,2])[0]))
```

```python
indexs = np.array(indexs)
sequences = df[['sequence']]
y = np.array(y)
sequences.to_csv(str(current_dir.parent / 'datasets/all_sequences.csv'), index=False)
```

```python
i_train, i_ex, y_train, y_ex = train_test_split(indexs, y, test_size=.2, random_state=42, stratify=y)
i_valid, i_test, y_valid, y_test = train_test_split(i_ex, y_ex, test_size=.5, random_state=42, stratify=y_ex)
```

```python
np.save(str(current_dir.parent / 'datasets/Second_Level/train_emb.npy'), i_train)
np.save(str(current_dir.parent / 'datasets/Second_Level/valid_emb.npy'), i_valid)
np.save(str(current_dir.parent / 'datasets/Second_Level/test_emb.npy'), i_test)
np.save(str(current_dir.parent / 'datasets/Second_Level/y_train.npy'), y_train)
np.save(str(current_dir.parent / 'datasets/Second_Level/y_valid.npy'), y_valid)
np.save(str(current_dir.parent / 'datasets/Second_Level/y_test.npy'), y_test)
```
