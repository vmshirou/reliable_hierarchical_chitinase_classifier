```python
import pandas as pd
from pathlib import Path
from sklearn.model_selection import train_test_split
```

```python
current_dir = Path.cwd().parent
```

```python
df_swiss_train = pd.read_json(str(current_dir / "datasets/SwissProt_2021_04/SwissProt_2021_04_train.json"))
df_swiss_train
```

```python
df_swiss_test = pd.read_json(str(current_dir / "datasets/SwissProt_2021_04/SwissProt_2021_04_test.json"))
df_swiss_test
```

```python
df_swiss_val = pd.read_json(str(current_dir / "datasets/SwissProt_2021_04/SwissProt_2021_04_valid.json"))
df_swiss_val
```

```python
df_18 = pd.read_csv(str(current_dir / 'datasets/GH18.csv'))
df_19 = pd.read_csv(str(current_dir / 'datasets/GH19.csv'))
```

```python
df_18["Familia"] = "GH18"
df_18
```

```python
df_19["Familia"] = "GH19"
df_19
```

```python
# Unir los DataFrames
df_combined = pd.concat([df_18, df_19], ignore_index=True)

df_combined
```

```python
df_swiss_all = pd.concat([df_swiss_train, df_swiss_test, df_swiss_val], ignore_index=True)
df_swiss_all
```

```python
len(df_swiss_all["sequence"].values)
```

```python
len(set(df_swiss_all["sequence"].values))
```

```python
# Buscar filas donde la columna 'sequence' se repita
duplicados = df_swiss_all[df_swiss_all.duplicated(subset=['sequence'], keep=False)]
duplicados
```

```python
final = df_swiss_all.drop_duplicates(subset=['sequence'], keep='first')
final
```

```python
final["ec_number"].value_counts()
```

```python
# Agregar la columna 'IsEnzime' con valores 1 si 'ec_number' es distinto de '0.0.0.0', 0 si no
final.loc[:, 'IsEnzime'] = final['ec_number'].apply(lambda x: 1 if x != '0.0.0.0' else 0)
final
```

```python
final["IsEnzime"].value_counts()
```

```python
# Dividir el DataFrame en 80% train y 20% restante (test + validación)
df_train, df_temp = train_test_split(final, test_size=0.1, stratify=final['IsEnzime'], random_state=42)

# Dividir el 10% restante en 10% test y 10% validación
df_test, df_val = train_test_split(df_temp, test_size=0.5, stratify=df_temp['IsEnzime'], random_state=42)

# Ver tamaños de los conjuntos
print(f'Tamaño del train set: {len(df_train)}')
print(f'Tamaño del test set: {len(df_test)}')
print(f'Tamaño del validation set: {len(df_val)}')
```

```python
# Contar cuántas instancias hay de cada clase en el conjunto de entrenamiento
train_counts = df_train['IsEnzime'].value_counts()

# Contar cuántas instancias hay de cada clase en el conjunto de prueba
test_counts = df_test['IsEnzime'].value_counts()

# Contar cuántas instancias hay de cada clase en el conjunto de validación
val_counts = df_val['IsEnzime'].value_counts()

print("Conteo en el conjunto de entrenamiento:")
print(train_counts)

print("\nConteo en el conjunto de prueba:")
print(test_counts)

print("\nConteo en el conjunto de validación:")
print(val_counts)
```

```python
df_train.to_csv(str(current_dir / 'datasets/Train/train.csv'), index=False)
df_test.to_csv(str(current_dir / 'datasets/Test/test.csv'), index=False)
df_val.to_csv(str(current_dir / 'datasets/Validation/valid.csv'), index=False)
```
