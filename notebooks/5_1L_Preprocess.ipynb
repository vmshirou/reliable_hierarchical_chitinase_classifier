```python
from pathlib import Path
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models, callbacks, optimizers, losses, metrics
from sklearn.preprocessing import StandardScaler
```

```python
current_dir = Path.cwd()
df_train_path = str(current_dir.parent / 'datasets/First Level/train_descriptores.csv')
df_valid_path = str(current_dir.parent / 'datasets/First Level/valid_descriptores.csv')
df_test_path = str(current_dir.parent / 'datasets/First Level/test_descriptores.csv')
train_emb_path = str(current_dir.parent / 'datasets/First Level/train_embeddings.npy')
valid_emb_path = str(current_dir.parent / 'datasets/First Level/valid_embeddings.npy')
test_emb_path = str(current_dir.parent / 'datasets/First Level/test_embeddings.npy')
```

```python
df = pd.read_csv(df_test_path)
sequences = df[['sequence']]
sequences.to_csv(str(current_dir.parent / 'datasets/First Level/test_sequences.csv'), index=False)
```

```python
df_train = pd.read_csv(df_train_path)
df_valid = pd.read_csv(df_valid_path)
df_test = pd.read_csv(df_test_path)
df_train.head(3)
```

```python
train_feat = df_train.iloc[:,4:].to_numpy()
valid_feat = df_valid.iloc[:,4:].to_numpy()
test_feat = df_test.iloc[:,4:].to_numpy()
y_train = df_train.iloc[:,3].to_numpy()
y_valid = df_valid.iloc[:,3].to_numpy()
y_test = df_test.iloc[:,3].to_numpy()
test_feat.shape
```

```python
np.save(str(current_dir.parent / 'datasets/First Level/train_feat.npy'), train_feat)
np.save(str(current_dir.parent / 'datasets/First Level/y_train.npy'), y_train)
np.save(str(current_dir.parent / 'datasets/First Level/valid_feat.npy'), valid_feat)
np.save(str(current_dir.parent / 'datasets/First Level/y_valid.npy'), y_valid)
np.save(str(current_dir.parent / 'datasets/First Level/test_feat.npy'), test_feat)
np.save(str(current_dir.parent / 'datasets/First Level/y_test.npy'), y_test)
```

```python
train_emb = np.load(train_emb_path)
valid_emb = np.load(valid_emb_path)
test_emb = np.load(test_emb_path)
```

```python
ptf_path = str(current_dir.parent / 'datasets/First Level/pre_processed/train_feat.npy')
pte_path = str(current_dir.parent / 'datasets/First Level/pre_processed/train_emb.npy')
pvf_path = str(current_dir.parent / 'datasets/First Level/pre_processed/valid_feat.npy')
pve_path = str(current_dir.parent / 'datasets/First Level/pre_processed/valid_emb.npy')
ptef_path = str(current_dir.parent / 'datasets/First Level/pre_processed/test_feat.npy')
ptee_path = str(current_dir.parent / 'datasets/First Level/pre_processed/test_emb.npy')
y_train_path = str(current_dir.parent / 'datasets/First Level/y_train.npy')
y_valid_path = str(current_dir.parent / 'datasets/First Level/y_valid.npy')
y_test_path = str(current_dir.parent / 'datasets/First Level/y_test.npy')
```

```python
import joblib
s_scaler_t = StandardScaler()
s_scaler_f = StandardScaler()
s_scaler_t.fit(train_emb)
s_scaler_f.fit(train_feat)
train_emb = s_scaler_t.transform(train_emb)
train_feat = s_scaler_f.transform(train_feat)
valid_emb = s_scaler_t.transform(valid_emb)
valid_feat = s_scaler_f.transform(valid_feat)
test_emb = s_scaler_t.transform(test_emb)
test_feat = s_scaler_f.transform(test_feat)
np.save(ptf_path, train_feat)
np.save(pte_path, train_emb)
np.save(pvf_path, valid_feat)
np.save(pve_path, valid_emb)
np.save(ptef_path, test_feat)
np.save(ptee_path, test_emb)
joblib.dump(s_scaler_f, str(current_dir.parent / 'datasets/First Level/s_scaler_f.pkl'))
joblib.dump(s_scaler_t, str(current_dir.parent / 'datasets/First Level/s_scaler_t.pkl'))
```
